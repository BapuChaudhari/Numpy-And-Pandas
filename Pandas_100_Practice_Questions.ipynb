### Q4. Perform arithmetic operations on a Series.
> 💡 **Hint:** Use `+`, `-`, `*`, `/`, `//`, `%`, `**` directly on the Series. Operations are **element-wise** and **index-aligned**.

---

```python
df=pd.Series([1,2,3,4,5,6,7,8],index=['a','b','c','d','e','f','g','h'])
print(df)

print('Division by 10')
print(df['a':'c']/10)

print('Addition of 10 \n')
print(df+10)

print('Subtraction of 10 \n')
print(df-10)

print('IntegerDivision by 2')
print(df['a':'c']//2)

print('Modulo by 2')
print(df['a':'c']%2)

print('Square of the series')
print(df**2)
```

### Q56. Detect outliers in a column.
> 💡 **Hint:** Use the **IQR method**: compute Q1, Q3, IQR = Q3–Q1. Outliers are values below `Q1 - 1.5*IQR` or above `Q3 + 1.5*IQR`. Alternatively use Z-score: `|z| > 3`.


---

```python
raw_df['Salary']=raw_df['Salary'].fillna(100000)
raw_df.iloc[8,2]=None
raw_df
```

```python
# Z-score
mean=raw_df['Salary'].mean()
print(mean)
std = raw_df['Salary'].std()
print(std)

z_score = (raw_df['Salary']-mean)/std
print(z_score >3)
print(z_score)
```

```python
Q1=raw_df['Salary'].quantile(.25)
print(Q1)
Q2=raw_df['Salary'].quantile(0.50)
print(Q2)
Q3=raw_df['Salary'].quantile(0.75)
print(Q3)
IQR=Q3-Q1
print(IQR)
outlier_lower = Q1 - 1.5*IQR
print(outlier_lower)
outlier_upper = Q3 + 1.5*IQR
print(outlier_upper)
```
