# Retail Sales Analytics
## Phase 2: Data Cleaning
### Perform checks such as:
- Remove duplicates 
- Handle missing values 
- Convert date columns 
- Fix inconsistent category names 
- Check for negative quantities 
- Check negative profits 
- Check impossible discounts 
- Verify unique IDs 


```python
import numpy as np
import pandas as pd
sales = pd.read_csv(r'D:\Data Analytics Projects\Retail Sales Analytics\Data\Raw Data\Sales_csv.csv')
sales.head()
```

Remove Duplicate rows

```python
sales.duplicated().sum()
# no duplicate rows found
```

Missing values

```python
print("Missing values per column:")
sales.isnull().sum()
```

Date Column Handeling

```python
sales['Order Date'] = pd.to_datetime(sales['Order Date'], format='%d-%m-%Y', errors='coerce')
sales['Ship Date'] = pd.to_datetime(sales['Ship Date'], format='%d-%m-%Y', errors='coerce')
```

```python
sales.sample()
```

Fixing inconsistent category names

```python
categorical_cols = [
    'Segment',
    'Country',
    'Region',
    'Category',
    'Sub-Category'
]

for col in categorical_cols:
    print(f"\n{col}")
    print(sales[col].value_counts())

for col in categorical_cols:
    sales[col] = sales[col].str.strip()

# All data seems to be consistant so no fixing needed.
```

Validate Numeric Columns for Invalid Negative Values

```python
print("Negative Values in dataset:")

print("In Quantity Column:", (sales['Quantity'] <= 0).sum())

print("In Sales Column:", (sales['Sales'] < 0).sum())

print("In Profit Column:", (sales['Profit'] < 0).sum())

# Profit has 1871 negative values
```

Checking Negative Profit

```python
sales[sales['Profit'] < 0]
```

Checking Discounts

```python
sales['Discount'].describe()
sales['Discount'].value_counts().sort_index()

sales[
    (sales['Discount'] < 0) |
    (sales['Discount'] > 1)
]

# Discount seems okay but some are on higher side like 70% and 80%
```

Verify Unique IDs

```python
sales['Customer ID'].nunique() # 793

sales['Order ID'].nunique()  # 5009

sales['Row ID'].is_unique  # True

sales[sales['Row ID'].duplicated()] # No Duplicate Row id


```

```python
clean_sales = sales.copy()
clean_sales.to_csv(r'D:\Data Analytics Projects\Retail Sales Analytics\Data\Cleaned Data\Clean_Sales_csv.csv', index=False)
```
