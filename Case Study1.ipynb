```python
import numpy as np 
import pandas as pd
```

```python
df = pd.read_csv(r"D:\DataSet\customer_data.csv")

print("Shape of DataSet:", df.shape)
print("\nData Types of each column:")
print(df.dtypes)
print("\nNull Values is DataSet")
print(pd.isnull(df).sum())
print("\n Description of dataset ")
print(df.describe)
print('check duplicate\n')
print(df.duplicated().sum())
df.head()
```

```python
# 	Average Purchase Value by age and gender.
df.groupby(['Gender','Age'])[['TotalAmountSpent']].mean()
```

```python
# 	Total Purchase Amount by location.
df.groupby(['Location'])[['TotalAmountSpent']].sum()
```

```python
#	Churn Rate and customer loyalty based on purchase frequency.
# customer loyalty based on purchase frequency.
df[df['TotalPurchases']>df['TotalPurchases'].mean()].shape

conditions=[
    df['TotalPurchases'] < 14,
    df['TotalPurchases'].between(14,16),
    df['TotalPurchases'] > 16
]
choices=['Low','Medium','High']

df["loyalty"] = np.select(conditions, choices, default="Unknown")
df
```

```python
df["LastPurchaseDate"]
```

```python
#	Churn Rate

df["LastPurchaseDate"] = pd.to_datetime(df["LastPurchaseDate"])
reference_date = pd.Timestamp("2024-12-31")

days_since_last_purchase = (
    reference_date - df["LastPurchaseDate"]
).dt.days

df["Churn"] = np.where(days_since_last_purchase > 360, "Yes", "No")
df
```

```python
#	Create personalized offers for the most valuable customers.
valuable_customers = df.sort_values('TotalAmountSpent',ascending=False)
valuable_customers[valuable_customers['TotalAmountSpent']>3000]
```
