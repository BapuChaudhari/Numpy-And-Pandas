Column names and Data types

Missing values 

```python
print("Missing Values in each Column")
print("-----------------------------")
print(sales.isnull().sum())
print("-----------------------------")
```

Duplicate rows 

```python
print("Duplicate Rows:", sales.duplicated().sum())
```

Numerical columns 

```python
print("\nNumerical Columns:")
print(sales.select_dtypes(include='number').columns)
```

Categorical columns 

```python
print("\nCategorical Columns:")
print(sales.select_dtypes(include=['object', 'string']).columns)

```

Memory usage 

```python
print("Memory Usage of each columns in bytes:")
print(sales.memory_usage(deep=True))
print("\nTotal Memory usage in MB:", sales.memory_usage(deep=True).sum() / (1024**2))

```
